# Gradient Descent: Engineering the Optimization Engine

When closed-form solutions (like the Normal Equation) become too computationally expensive for massive datasets, we rely on Gradient Descent. It is an iterative optimization algorithm that uses partial derivatives to navigate the "Loss Landscape" and find the minimum error.

# Phase1: i am creating a data then find m and b with the scikit OLS method.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# generate data
X, y = make_regression(n_samples=100, n_features=1, noise=20, random_state=42)
X_flat = X.ravel()

ols = LinearRegression()
ols.fit(X, y)

target_m = ols.coef_[0]
target_b = ols.intercept_
target_r2 = r2_score(y, ols.predict(X))

print(f"--- THE TARGET (Scikit-Learn) ---")
print(f"Optimal Slope (m):     {target_m:.4f}")
print(f"Optimal Intercept (b): {target_b:.4f}")
print(f"Target R2 Score:       {target_r2:.4f}")
```

# Phase-2: 1D gradient Descent ( isolating the intercept)
To prove the math works, we lock the slope (m) to the optimal Scikit-Learn value. We initialize our intercept (b) at a terrible guess, and force the algorithm to step down the 2D parabola to find the optimal b.

We will use matplotlib.animation to generate an HTML5 video directly inside your Colab notebook, showing both the regression line fitting the data and the error sliding down the loss parabola.

```python
import matplotlib.animation as animation
from IPython.display import HTML
```

```python
class GradientDescent1D:
  def __init__(self, m_fixed, learning_rate=0.1, epochs=50):
    self.m = m_fixed
    self.b = -100   # start with a guess
    self.lr = learning_rate
    self.epochs = epochs

    # logging for animation
    self.b_history = []
    self.loss_history = []

  def fit(self, X, y):
    n = len(X)
    for i in range(self.epochs):
      y_pred = self.m * X + self.b
      loss = np.mean((y - y_pred) ** 2)

      self.b_history.append(self.b)
      self.loss_history.append(loss)

      # derivative of MSE w.r.t b
      dl_db = -2 * np.mean(y - y_pred)

      # update rule
      self.b = self.b - (self.lr * dl_db)
```

```python
# Run the Custom 1D Engine
gd_1d = GradientDescent1D(m_fixed=target_m, learning_rate=0.1, epochs=40)
gd_1d.fit(X_flat, y)

print(f"1D Custom Intercept: {gd_1d.b:.4f} \nOLR previous intercept: {target_b}\n(Matches Target: {np.isclose(gd_1d.b, target_b, atol=0.01)})")
```

```python
# Animation logic
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14,5))
plt.close() # prevents static plot from rendering

def update_1d(frame):
    ax1.clear()
    ax2.clear()

    current_b = gd_1d.b_history[frame]
    current_loss = gd_1d.loss_history[frame]

    # Plot 1: Regression Line Fitting
    ax1.scatter(X_flat, y, color='lightgray')
    ax1.plot(X_flat, target_m * X_flat + current_b, color='red', linewidth=2)
    ax1.set_title(f"Epoch {frame}: Fitting the Line")
    ax1.set_ylim(-150, 150)

    # Plot 2: Sliding Down the Parabola
    b_range = np.linspace(-120, 120, 100)
    loss_curve = [np.mean((y - (target_m * X_flat + b_val))**2) for b_val in b_range]

    ax2.plot(b_range, loss_curve, color='blue')
    ax2.scatter(current_b, current_loss, color='red', s=100, zorder=5)
    ax2.set_title(f"Loss Function: MSE = {current_loss:.2f}")
    ax2.set_xlabel("Intercept (b)")

anim = animation.FuncAnimation(fig, update_1d, frames=len(gd_1d.b_history), interval=200)
HTML(anim.to_jshtml())
```

# Phase3: 2D Gradient Descent
Now i am practicing with both variables. the loss function is not now 2-D parabola but a 3-D bowl.

```python
class GradientDescent2D:
  def __init__(self, learning_rate=0.1, epochs=50):
    self.m = -200 # terrible guess
    self.b = -150 # terrible guess
    self.lr = learning_rate
    self.epochs = epochs

    self.m_history = []
    self.b_history = []
    self.loss_history = []

  def fit(self, X, y):
    n = len(X)
    for i in range(self.epochs):
      y_pred = self.m * X + self.b
      loss = np.mean((y - y_pred) ** 2)

      self.m_history.append(self.m)
      self.b_history.append(self.b)
      self.loss_history.append(loss)

      # partial derivatives
      dl_dm = -2 * np.mean(X * (y-y_pred))
      dl_db = -2 * np.mean(y - y_pred)

      # simultaneous update
      self.m = self.m - (self.lr * dl_dm)
      self.b = self.b - (self.lr * dl_db)
```

```python
# custom 2D engine
gd_2d = GradientDescent2D()
gd_2d.fit(X_flat, y)

print(f"2D Custom Slope:     {gd_2d.m:.4f}")
print(f"2D Custom Intercept: {gd_2d.b:.4f}")
```

```python
# Animation Logic: Contour Path
fig, ax = plt.subplots(figsize=(8, 6))
plt.close()

# Generate the 3D bowl data for the contour
m_grid = np.linspace(-120, 120, 100)
b_grid = np.linspace(-120, 120, 100)
M, B = np.meshgrid(m_grid, b_grid)
Z = np.zeros_like(M)
for i in range(100):
    for j in range(100):
        y_pred_grid = M[i, j] * X_flat + B[i, j]
        Z[i, j] = np.mean((y - y_pred_grid)**2)

def update_2d(frame):
    ax.clear()

    # Plot the Contour (The topographical map of the Loss Bowl)
    contour = ax.contour(M, B, Z, levels=np.logspace(0, 5, 35), cmap='viridis', alpha=0.5)

    # Plot the path taken by Gradient Descent up to the current frame
    ax.plot(gd_2d.m_history[:frame], gd_2d.b_history[:frame], color='red', marker='o', markersize=4)

    # Plot the Target Center
    ax.scatter(target_m, target_b, color='black', marker='*', s=200, label='Optimal (Scikit-Learn)')

    ax.set_title(f"Epoch {frame}: Traversing the Loss Landscape")
    ax.set_xlabel("Slope (m)")
    ax.set_ylabel("Intercept (b)")
    ax.legend(loc='upper left')

anim_2d = animation.FuncAnimation(fig, update_2d, frames=len(gd_2d.m_history), interval=150)
HTML(anim_2d.to_jshtml())
```
