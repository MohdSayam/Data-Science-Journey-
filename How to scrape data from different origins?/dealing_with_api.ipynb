```python
import pandas as pd
import requests
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
```

```python
print(response.status_code)
```

```python
# converting it to json
data = response.json()
print(type(data))

# here api is returning list of dictionaries so type is showing list
```

```python
print(data[0])
```

```python
df = pd.DataFrame(data)
df.head()
```

# Now we can perform various pandas operations on it like:
- sorting
- filtering
- string operations etc

```python
print(df[df["userId"]==1])
```

```python
df['title'].str.upper()
```

```python
# we can create new columns as it is a dataframe
df['title_length'] = df['title'].str.len()
df
```

# we can even visualize this data now

```python
import matplotlib.pyplot as plt

df["userId"].value_counts().plot(kind="bar")

plt.title("Posts Per User")

plt.show()
```
