# Working with CSV Files in Pandas

This notebook provides a detailed guide on how to effectively work with CSV (Comma Separated Values) files using the powerful pandas library in Python. We'll cover everything from basic loading to handling common issues and leveraging advanced parameters.

````python
# First, let's create some dummy CSV files for demonstration purposes
# This helps ensure the examples are reproducible without external files.

import pandas as pd
import numpy as np
import os

# Dummy file 1: aug_train.csv for basic reading
data_basic = {
    'enrollee_id': [1, 2, 3, 4, 5],
    'city': ['Mumbai', 'Bangalore', 'Delhi', 'Kolkata', 'Chennai'],
    'gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
    'relevent_experience': ['Has relevent experience', 'No relevent experience', 'Has relevent experience', 'Has relevent experience', 'No relevent experience'],
    'education_level': ['Graduate', 'Masters', 'Graduate', 'Phd', 'Graduate'],
    'experience': [2, 5, 8, 1, 10],
    'target': [0, 1, 0, 1, 0]
}
df_basic = pd.DataFrame(data_basic)
df_basic.to_csv('aug_train.csv', index=False)
print('Created aug_train.csv')

# Dummy file 2: data.tsv for `sep` parameter
data_tsv = {
    'col1': [10, 20, 30],
    'col2': ['A', 'B', 'C'],
    'col3': [1.1, 2.2, 3.3]
}
df_tsv = pd.DataFrame(data_tsv)
df_tsv.to_csv('data.tsv', index=False, sep='\t')
print('Created data.tsv')

# Dummy file 3: data_no_header.csv for `header` parameter
data_no_header = [
    [1, 'Apple', 100],
    [2, 'Banana', 150],
    [3, 'Orange', 200]
]
with open('data_no_header.csv', 'w') as f:
    for row in data_no_header:
        f.write(','.join(map(str, row)) + '\n')
print('Created data_no_header.csv')

# Dummy file 4: data_skip_rows.csv for `skiprows` parameter
data_skip_rows = [
    '# This is a comment',
    '# Another comment line',
    'ID,Product,Price',
    '1,Desk,50',
    '2,Chair,30'
]
with open('data_skip_rows.csv', 'w') as f:
    for line in data_skip_rows:
        f.write(line + '\n')
print('Created data_skip_rows.csv')

# Dummy file 5: data_dates.csv for `parse_dates` and `na_values`
data_dates = {
    'OrderDate': ['2023-01-01', '2023-01-02', '2023-01-XX', '2023-01-04'],
    'Amount': [100, 150, 200, np.nan]
}
df_dates = pd.DataFrame(data_dates)
df_dates.to_csv('data_dates.csv', index=False)
print('Created data_dates.csv')

# Dummy file 6: data_latin1.csv for `encoding`
data_latin1_content = 'Name,Description\nJosé,Camión\nMaria,Avión'
with open('data_latin1.csv', 'w', encoding='latin-1') as f:
    f.write(data_latin1_content)
print('Created data_latin1.csv')


print('\nAll dummy CSVs created successfully!')
````

## Section 1: Introduction to CSV Files

### Theory: What are CSVs and Why Pandas?

*   **CSV (Comma Separated Values)** files are plain text files that use commas to separate values. Each line in the file represents a row of data, and within each row, commas separate the individual data fields (columns).
*   **Why datasets use CSV**: They are widely used for data exchange because of their simplicity and human-readability. Almost any spreadsheet program, database, or data analysis tool can import and export data in CSV format.
*   **Why Pandas is used**: The `pandas` library in Python provides highly optimized and flexible data structures, particularly the `DataFrame`, which is perfect for handling tabular data like that found in CSV files. `pandas` simplifies the process of reading, writing, and manipulating CSV data with powerful functions like `pd.read_csv()`.

## Section 4: Important `pd.read_csv()` Parameters

Understanding `pd.read_csv()` parameters is crucial for handling diverse CSV file formats and optimizing data loading. Instead of memorizing them, focus on the 'problem solved' by each parameter.

### Parameter: `sep`

#### Theory

*   **What it does**: The `sep` parameter specifies the delimiter (separator) used in the file to separate values. By default, `pd.read_csv()` assumes a comma (`,`) as the separator.
*   **Why it is used**: Many tabular data files are not strictly comma-separated. They might use tabs (`\t`), semicolons (`;`), pipes (`|`), or other characters to delimit fields. When the default comma delimiter fails, `sep` allows you to specify the correct one.
*   **Common Mistakes**: Not specifying `sep` when the file uses a non-comma delimiter. This often results in a DataFrame with a single column containing all the data merged into one string per row.

#### Syntax

```python
pd.read_csv("file.csv", sep="delimiter_character")
```

Common values for `delimiter_character` include `','` (default), `'\t'` for tab-separated values (TSV), `';'`, or `'|'`.

#### Example: Reading a Tab-Separated File

````python
import pandas as pd

# Try to read 'data.tsv' without specifying the separator (will likely fail to parse correctly)
print("Attempting to read data.tsv without 'sep':")
df_wrong_sep = pd.read_csv("data.tsv")
display(df_wrong_sep.head())

# Now, read 'data.tsv' with the correct tab separator
print("\nReading data.tsv with sep='\\t':")
df_tsv = pd.read_csv("data.tsv", sep='\t')
display(df_tsv.head())
````

#### Output Understanding

*   When `sep` is not specified for a tab-separated file, pandas might interpret the entire row as a single column, as seen in `df_wrong_sep`.
*   By setting `sep='\t'`, pandas correctly identifies the tab as the delimiter and parses the data into three distinct columns (`col1`, `col2`, `col3`) as intended.

### Parameter: `index_col`

#### Theory

*   **What it does**: The `index_col` parameter specifies which column (or columns) from your CSV file should be used as the DataFrame's row labels (index). By default, if not specified, pandas creates a new integer-based index starting from 0.
*   **Why it is used**: Often, your dataset has a natural identifier (e.g., 'ID', 'Date', 'Product Code') that would serve better as the DataFrame's index than a generic numerical range. Setting a meaningful index can make data selection, alignment, and analysis more intuitive and efficient.
*   **Common Mistakes**: Not setting an appropriate `index_col` when a unique identifier column exists, leading to a redundant column in your DataFrame and a generic index.

#### Syntax

```python
pd.read_csv("file.csv", index_col="column_name")
# or by position
pd.read_csv("file.csv", index_col=integer_position)
# for a multi-index
pd.read_csv("file.csv", index_col=[column_name_1, column_name_2])
```

You can provide the column name as a string or its zero-based integer position. For a hierarchical (MultiIndex) index, pass a list of column names or positions.

#### Example: Setting `enrollee_id` as Index

````python
import pandas as pd

# Read 'aug_train.csv' without specifying an index column (default behavior)
print("DataFrame without specified index_col:")
df_no_index = pd.read_csv('aug_train.csv')
display(df_no_index.head())

# Read 'aug_train.csv' and set 'enrollee_id' as the index
print("\nDataFrame with 'enrollee_id' as index_col:")
df_with_index = pd.read_csv('aug_train.csv', index_col='enrollee_id')
display(df_with_index.head())
````

#### Output Understanding

*   In `df_no_index`, pandas automatically assigned a default integer index (0, 1, 2, ...), and `enrollee_id` remains a regular column.
*   In `df_with_index`, the `enrollee_id` column has been promoted to the DataFrame's index, meaning it now serves as the unique label for each row. This makes it easier to select specific rows by `enrollee_id`.

### Parameter: `header`

#### Theory

*   **What it does**: The `header` parameter tells `pd.read_csv()` which row (if any) contains the column names. By default, `header=0`, meaning the first row (index 0) of the file is treated as the column names.
*   **Why it is used**: Some CSV files might not have a header row at all, or the header information could be on a different row (e.g., after some metadata). Correctly identifying the header ensures your DataFrame has proper column names.
*   **Common Mistakes**: Not setting `header=None` for files without a header, which leads to the first data row being incorrectly used as column names. Conversely, leaving `header=0` for files where the header is on a different line can cause important data to be lost or misidentified as column names.

#### Syntax

```python
pd.read_csv("file.csv", header=None) # No header
pd.read_csv("file.csv", header=integer_row_index) # Header on a specific row
```

Specify `None` if the file has no header. Use an integer (0-based index) to indicate the row number where the header is located.

#### Example: Reading a CSV Without a Header

````python
import pandas as pd

# Read 'data_no_header.csv' without specifying header (default header=0)
print("DataFrame from 'data_no_header.csv' without specifying header (incorrect):")
df_no_header_wrong = pd.read_csv('data_no_header.csv')
display(df_no_header_wrong.head())

# Read 'data_no_header.csv' correctly by setting header=None
print("\nDataFrame from 'data_no_header.csv' with header=None (correct):")
df_no_header_correct = pd.read_csv('data_no_header.csv', header=None)
display(df_no_header_correct.head())
````

#### Output Understanding

*   When `header` is not specified (`df_no_header_wrong`), pandas assumes the first row (`1,Apple,100`) contains column names, leading to `1`, `Apple`, `100` being used as column headers and the actual data starting from the second row.
*   By setting `header=None` (`df_no_header_correct`), pandas understands there are no explicit headers. It assigns default numerical column names (0, 1, 2) and correctly treats all rows as data.

### Parameter: `usecols`

#### Theory

*   **What it does**: The `usecols` parameter allows you to specify a subset of columns to load from the CSV file. Instead of loading all columns, pandas will only read and return the ones you list.
*   **Why it is used**: This parameter is particularly useful for working with very large datasets where loading all columns might consume excessive memory or be unnecessary for your analysis. It helps in optimizing memory usage and speeding up the loading process.
*   **Common Mistakes**: Providing incorrect column names, or forgetting that columns are 0-indexed if specifying by position.

#### Syntax

```python
pd.read_csv("file.csv", usecols=['col_name_1', 'col_name_2'])
# or by position
pd.read_csv("file.csv", usecols=[0, 2, 4])
```

You can pass a list of column names (strings) or a list of integer column positions (0-based) to `usecols`.

#### Example: Loading Specific Columns

````python
import pandas as pd

# Load 'aug_train.csv' but only include 'enrollee_id', 'city', and 'education_level' columns
print("DataFrame loaded with 'enrollee_id', 'city', 'education_level' columns:")
df_selected_cols = pd.read_csv('aug_train.csv', usecols=['enrollee_id', 'city', 'education_level'])
display(df_selected_cols.head())

# Load the same columns using their integer positions (0-indexed)
# enrollee_id (0), city (1), gender (2), relevent_experience (3), education_level (4), experience (5), target (6)
print("\nDataFrame loaded with columns at positions 0, 1, 4:")
df_selected_cols_by_pos = pd.read_csv('aug_train.csv', usecols=[0, 1, 4])
display(df_selected_cols_by_pos.head())
````

#### Output Understanding

*   Instead of all 7 columns, both `df_selected_cols` and `df_selected_cols_by_pos` only contain the specified three columns (`enrollee_id`, `city`, `education_level`).
*   This demonstrates how `usecols` effectively filters the data during the loading process, reducing memory footprint and potentially speeding up file reading for very wide datasets.

### Parameter: `nrows`

#### Theory

*   **What it does**: The `nrows` parameter specifies the number of rows to read from the beginning of the CSV file. After reading this many rows, pandas stops parsing the file.
*   **Why it is used**: This is invaluable for dealing with very large files, allowing you to quickly inspect the file's structure, column names, and data types without loading the entire dataset into memory. It's perfect for testing your `read_csv` parameters before committing to a full load.
*   **Common Mistakes**: Forgetting that `nrows` reads from the *beginning* of the file, so if your header is on row 5, setting `nrows=1` will only read one row *before* the header.

#### Syntax

```python
pd.read_csv("file.csv", nrows=integer_count)
```

`integer_count` is the number of data rows to read. Note that if a `header` is specified, the `nrows` count is *after* the header row (i.e., it counts actual data rows).

#### Example: Reading Only a Few Rows

````python
import pandas as pd

# Read only the first 3 data rows from 'aug_train.csv'
print("DataFrame loaded with only 3 data rows using nrows=3:")
df_nrows = pd.read_csv('aug_train.csv', nrows=3)
display(df_nrows.head())

print(f"Shape of the DataFrame: {df_nrows.shape}")
````

#### Output Understanding

*   The resulting DataFrame `df_nrows` contains only 3 data rows, as specified by `nrows=3`. The header was still correctly identified and used for column names.
*   The shape `(3, 7)` confirms that only 3 rows and all 7 columns were loaded. This is efficient for quick checks on large files.

### Parameter: `skiprows`

#### Theory

*   **What it does**: The `skiprows` parameter allows you to skip rows at the beginning of the file when reading it. You can specify a number of rows to skip, or a list of specific row numbers (0-indexed) to skip.
*   **Why it is used**: This is useful for CSV files that contain metadata, comments, or header information that you don't want to include in your DataFrame, located *before* the actual data or the column names. It's often used in conjunction with `header` if the header isn't on the very first row.
*   **Important Notes**: When `skiprows` is an integer, it skips that many rows *from the top of the file*. If it's a list of integers, it skips those specific row numbers.
*   **Common Mistakes**: Confusing `skiprows` (which skips rows *before* interpretation) with `header` (which defines which row *is* the header). They can be used together, but their effects are distinct.

#### Syntax

```python
pd.read_csv("file.csv", skiprows=integer_count)
# or
pd.read_csv("file.csv", skiprows=[row_index_1, row_index_2, ...])
```

`integer_count` skips the first N rows. A list of `row_index` (0-based) skips specific rows.

#### Example: Skipping Initial Rows

````python
import pandas as pd

# The 'data_skip_rows.csv' file has comments on the first two lines, then the header.
# Content:
# # This is a comment
# # Another comment line
# ID,Product,Price
# 1,Desk,50
# 2,Chair,30

# Attempt to read without skiprows (will incorrectly treat comments as data/header)
print("Attempting to read 'data_skip_rows.csv' without skiprows:")
df_skip_wrong = pd.read_csv('data_skip_rows.csv')
display(df_skip_wrong.head())

# Correctly skip the first two comment rows (index 0 and 1) and let pandas find the header on the third row (index 2)
print("\nReading 'data_skip_rows.csv' by skipping 2 rows:")
df_skiprows_int = pd.read_csv('data_skip_rows.csv', skiprows=2)
display(df_skiprows_int.head())

# Alternatively, skip specific row indices (e.g., first two rows)
print("\nReading 'data_skip_rows.csv' by skipping rows [0, 1]:")
df_skiprows_list = pd.read_csv('data_skip_rows.csv', skiprows=[0, 1])
display(df_skiprows_list.head())
````

### Parameter: `dtype`

#### Theory

*   **What it does**: The `dtype` parameter allows you to explicitly specify the data type for one or more columns when loading a CSV file. By default, pandas tries to infer the most appropriate data type for each column (e.g., `int64`, `float64`, `object`).
*   **Why it is used**: Explicitly setting data types is crucial for several reasons:
    *   **Memory Optimization**: Using smaller, more precise data types (e.g., `int8` instead of `int64` if values fit) can significantly reduce memory usage for large DataFrames.
    *   **Performance**: Operations on correctly typed data are often faster.
    *   **Data Integrity**: Prevents pandas from inferring an incorrect type (e.g., treating a column of numerical IDs as `float` because of a single `NaN`, or as `object` because of a mixed type).
    *   **Problem Solved**: Memory optimization, incorrect type inference, ensuring data integrity.

#### Syntax

```python
pd.read_csv("file.csv", dtype={'column_name': 'data_type'})
# or for multiple columns
pd.read_csv("file.csv", dtype={'col_A': 'int32', 'col_B': 'float16'})
```

Provide a dictionary where keys are column names (strings) and values are the desired data types (e.g., `'int'`, `'float'`, `'str'`, `'bool'`, `'category'`, `'datetime64[ns]'`).

### Parameter: `converters`

#### Theory

*   **What it does**: The `converters` parameter allows you to specify a function to apply to the values in a particular column (or columns) *before* pandas attempts to infer its `dtype` or load it into the DataFrame. This gives you fine-grained control over how individual cell values are parsed.
*   **Why it is used**: This is extremely powerful for custom data cleaning or transformation during the loading phase. For example, you might need to:
    *   Remove unwanted characters (e.g., currency symbols, percent signs).
    *   Apply a custom parsing logic for ambiguous date/time formats that `parse_dates` can't handle.
    *   Convert specific text values into numerical representations based on custom rules.
    *   Handle complex missing value patterns (though `na_values` is often simpler for this).
*   **Important Notes**: Converters are functions that take a single string argument (the cell's value) and return the transformed value. They are applied *before* `dtype` inference, and also *before* `na_values` are applied.
*   **Common Mistakes**: Overusing converters for simple tasks that `dtype`, `parse_dates`, or `na_values` can handle more efficiently. Not handling potential errors within the converter function (e.g., `ValueError` if attempting to convert a non-numeric string to a number).

#### Syntax

```python
pd.read_csv("file.csv", converters={'column_name': function_to_apply})
```

`converters` takes a dictionary where keys are column names (strings) and values are functions. Each function will receive the string value of a cell in that column and should return the converted value.

#### Example: Custom Conversion for 'Amount' Column

````python
import pandas as pd
import numpy as np

# Create a dummy CSV with a column needing custom cleaning
data_converters_content = (
    'Item,Price,Status\n'
    'Laptop,$1200.50,In Stock\n'
    'Mouse,$25.99,Out of Stock\n'
    'Keyboard,$75.00,Backorder\n'
    'Monitor,N/A,In Stock'
)
with open('data_converters.csv', 'w') as f:
    f.write(data_converters_content)

# Define a custom converter function for the 'Price' column
def clean_price(price_str):
    if price_str.startswith('$'):
        price_str = price_str[1:] # Remove '$'
    try:
        return float(price_str)
    except ValueError:
        return np.nan # Return NaN if conversion fails (e.g., for 'N/A')

# Read the CSV using the converter
print("DataFrame with 'Price' column cleaned using a converter:")
df_converted = pd.read_csv('data_converters.csv', converters={'Price': clean_price})
print(df_converted.dtypes)
display(df_converted.head())

# Clean up dummy file
os.remove('data_converters.csv')
````

### Parameter: `chunksize`

#### Theory

*   **What it does**: The `chunksize` parameter allows `pd.read_csv()` to read the file in smaller, manageable pieces (chunks) rather than loading the entire file into memory at once. When `chunksize` is specified, `read_csv()` returns an iterator (`TextFileReader` object), which you can then iterate over to process each chunk.
*   **Why it is used**: This is a critical parameter for handling **huge datasets** that exceed available RAM. Instead of failing due to memory errors, you can process the data chunk by chunk, performing aggregations, transformations, or writing to a database incrementally.
*   **Problem Solved**: Huge datasets, memory optimization.
*   **Common Mistakes**: Expecting a `DataFrame` directly when `chunksize` is used; it returns an iterator. Forgetting to explicitly combine or aggregate chunks if a single DataFrame is eventually desired.

#### Syntax

```python
csv_reader = pd.read_csv("file.csv", chunksize=integer_size)
for chunk in csv_reader:
    # Process each chunk (which is a DataFrame)
    pass
```

`integer_size` specifies the number of rows each chunk (DataFrame) will contain.

#### Example: Reading a Large File in Chunks

````python
import pandas as pd
import os

# Create a large dummy CSV file (e.g., 10,000 rows for demonstration)
num_rows = 10000
df_large_dummy = pd.DataFrame({
    'col1': range(num_rows),
    'col2': [f'val_{i % 10}' for i in range(num_rows)],
    'col3': [float(i) * 1.5 for i in range(num_rows)]
})
df_large_dummy.to_csv('large_data.csv', index=False)
print(f"Created 'large_data.csv' with {num_rows} rows.")

# Read 'large_data.csv' in chunks of 2000 rows
chunk_size = 2000
csv_reader = pd.read_csv('large_data.csv', chunksize=chunk_size)

print(f"\nReading 'large_data.csv' in chunks of {chunk_size} rows:")

# Process each chunk
all_chunks = []
for i, chunk in enumerate(csv_reader):
    print(f"  Processing chunk {i+1}, shape: {chunk.shape}")
    # Example: Store first 2 rows of each chunk
    if i == 0:
        print("    First 2 rows of first chunk:")
        display(chunk.head(2))
    all_chunks.append(chunk)

# Optionally, concatenate all chunks back into a single DataFrame
final_df = pd.concat(all_chunks)
print(f"\nConcatenated all chunks into a single DataFrame of shape: {final_df.shape}")

# Clean up dummy file
os.remove('large_data.csv')
````

#### Output Understanding

*   Instead of a single DataFrame, `pd.read_csv()` with `chunksize` returns a `TextFileReader` object, which is an iterator.
*   The loop demonstrates iterating through this object, with each `chunk` being a pandas DataFrame of the specified `chunk_size` (or smaller for the last chunk).
*   This approach allows you to perform operations on smaller parts of the data, which is critical for memory management with large files. Finally, all processed chunks can be optionally recombined using `pd.concat`.

### Parameter: `names`

#### Theory

*   **What it does**: The `names` parameter allows you to assign a list of column names to your DataFrame. This is particularly useful when your CSV file does not have a header row, or when you want to provide custom names that differ from those in the file's header.
*   **Why it is used**:
    *   **Files without headers**: When `header=None`, pandas assigns default integer column names (0, 1, 2...). `names` lets you provide meaningful names.
    *   **Renaming columns on load**: You might want to standardize column names (e.g., lowercase, replace spaces with underscores) directly during loading, instead of renaming them afterward.
*   **Important Notes**:
    *   If `names` is provided, pandas assumes the file *does not* have a header row unless `header` is also explicitly set to an integer (e.g., `header=0`). If both `names` and `header=0` are provided, the `names` list will *replace* the existing header from the file. If `header=None` and `names` is provided, the first row of the file is treated as data.
    *   The number of names in the list should match the number of columns in the file.
*   **Common Mistakes**: Providing `names` without setting `header=None` for files that genuinely lack a header, which can cause the actual first data row to be mistaken for the header and then overwritten by the `names` list.

#### Syntax

```python
pd.read_csv("file.csv", header=None, names=['colA', 'colB', 'colC'])
# or to replace an existing header
pd.read_csv("file.csv", header=0, names=['new_colA', 'new_colB', 'new_colC'])
```

`names` expects a list of strings, where each string is the desired name for a column, in order.

#### Example: Assigning Custom Column Names

### Parameter: `on_bad_lines` (formerly `error_bad_lines`)

#### Theory

*   **What it does**: The `on_bad_lines` parameter controls how pandas handles rows that have too many or too few fields compared to the expected number of columns. This parameter replaced the deprecated `error_bad_lines` in pandas 1.3.0 and later.
    *   `'error'` (default): Raises an exception (like `ParserError`) if a bad line is encountered.
    *   `'warn'`: Skips bad lines and prints a warning message. Processing continues.
    *   `'skip'`: Skips bad lines silently without raising an error or printing a warning.
*   **Why it is used**: CSV files, especially those from manual entry or legacy systems, can often be malformed. A line might contain an extra comma, or be missing values, leading to a different number of fields than the header. This parameter allows you to gracefully handle such inconsistencies without crashing your script or manually cleaning the file.
*   **Problem Solved**: Handling malformed rows, preventing `ParserError`.
*   **Common Mistakes**: Ignoring `ParserError` without using `on_bad_lines`, leading to script crashes. Using `'skip'` blindly, which might lead to silent data loss if many important rows are malformed.

#### Syntax

```python
pd.read_csv("file.csv", on_bad_lines='error') # Default
pd.read_csv("file.csv", on_bad_lines='warn')
pd.read_csv("file.csv", on_bad_lines='skip')
```

`on_bad_lines` accepts a string value: `'error'`, `'warn'`, or `'skip'`.

#### Example: Handling Malformed Lines

````python
import pandas as pd
import os

# Create a dummy CSV with some bad lines
data_bad_lines_content = (
    'ColA,ColB,ColC\n'
    '1,2,3\n'
    '4,5,6,7\n'   # Too many fields
    '8,9\n'      # Too few fields
    '10,11,12'
)
with open('data_bad_lines.csv', 'w') as f:
    f.write(data_bad_lines_content)

# Attempt to read with default 'on_bad_lines' ('error')
print("Attempting to read 'data_bad_lines.csv' with default 'error' behavior:")
try:
    df_bad_error = pd.read_csv('data_bad_lines.csv')
    display(df_bad_error.head())
except pd.errors.ParserError as e:
    print(f"Caught expected ParserError: {e}")

# Read with 'on_bad_lines' set to 'warn'
print("\nReading 'data_bad_lines.csv' with 'on_bad_lines='warn':")
df_bad_warn = pd.read_csv('data_bad_lines.csv', on_bad_lines='warn')
display(df_bad_warn.head())

# Read with 'on_bad_lines' set to 'skip'
print("\nReading 'data_bad_lines.csv' with 'on_bad_lines='skip':")
df_bad_skip = pd.read_csv('data_bad_lines.csv', on_bad_lines='skip')
display(df_bad_skip.head())

# Clean up dummy file
os.remove('data_bad_lines.csv')
````

#### Output Understanding

*   When `on_bad_lines='error'` (default), pandas stops processing and raises a `ParserError` upon encountering the first malformed line. This is good for strict data quality control.
*   With `on_bad_lines='warn'`, pandas skips the bad lines, prints a warning message to standard error (which you'll see in the output), and returns a DataFrame with only the correctly parsed rows. This is useful for identifying issues while still loading available data.
*   Using `on_bad_lines='skip'` silently drops the malformed rows without any error or warning. This can be dangerous as it hides data quality issues, but might be acceptable if you are confident bad lines are inconsequential.

## Section 5: Common CSV Problems and Solutions

This section summarizes common issues encountered when reading CSV files and directly maps them to the `pd.read_csv()` parameters that solve them. This approach focuses on problem-solving rather than rote memorization.

### Problem 1: `ParserError` or Data in a Single Column

#### Problem

When you load a CSV, all data appears in a single column, or you get a `ParserError` like 'Expected N fields in line Y, saw Z fields'.

#### Solution

This typically means pandas couldn't correctly identify the delimiter between columns. The default is a comma (`,`), but your file might use a semicolon (`;`), tab (`\t`), or pipe (`|`).

**Parameter to use**: `sep`

```python
# For semicolon-separated values
df = pd.read_csv("my_data.csv", sep=';')

# For tab-separated values (TSV)
df = pd.read_csv("my_data.tsv", sep='\t')
```

### Problem 2: Wrong Column Names or Missing Header

#### Problem

Your DataFrame's column names are incorrect (e.g., the first data row became the header), or the file doesn't have a header at all, or the header is on a line other than the first.

#### Solution

Adjust the `header` parameter to point to the correct row containing column names, or set it to `None` if no header exists. If no header, use `names` to assign them.

**Parameters to use**: `header`, `names`, `skiprows`

```python
# File has no header, assign custom names
df = pd.read_csv("data_no_header.csv", header=None, names=['col1', 'col2', 'col3'])

# Header is on the 3rd row (index 2)
df = pd.read_csv("data_with_header_row3.csv", header=2)

# Header on row 11, skip 10 descriptive rows above it
df = pd.read_csv("data_with_meta.csv", skiprows=10, header=0) # header=0 refers to the 0-indexed row *after* skipping
```

### Problem 3: `UnicodeDecodeError` or 'Weird Characters'

#### Problem

You get a `UnicodeDecodeError` or special characters (like `é`, `ñ`, `ä`) appear as gibberish (e.g., `Ã©`, `Ã±`).

#### Solution

The file's character encoding is different from pandas' default (`UTF-8`). You need to specify the correct encoding.

**Parameter to use**: `encoding`

```python
# Try common encodings for non-UTF-8 files
df = pd.read_csv("my_file.csv", encoding='latin-1')

# Another common encoding, especially for Windows-created files
df = pd.read_csv("my_file.csv", encoding='cp1252')

# If you don't know the encoding, use a library like `chardet` to guess it:
# import chardet

```

### Problem 4: Dates are Loaded as Strings (`object` Dtype)

#### Problem

Columns that clearly contain dates (e.g., '2023-01-15') are loaded as generic `object` (string) type, preventing date-based operations.

#### Solution

Tell pandas which columns should be parsed as `datetime` objects.

**Parameter to use**: `parse_dates`

```python
# Parse a single column
df = pd.read_csv("sales.csv", parse_dates=['OrderDate'])

# Parse multiple columns
df = pd.read_csv("events.csv", parse_dates=['StartDate', 'EndDate'])

# Combine 'Year', 'Month', 'Day' columns into one datetime column
df = pd.read_csv("split_dates.csv", parse_dates=[['Year', 'Month', 'Day']])
```

### Problem 5: Custom Missing Value Indicators

#### Problem

Your data uses custom strings (e.g., `'NA'`, `'?'`, `'-'`) or numbers (e.g., `-999`) to represent missing values, but pandas loads them as regular data, potentially forcing numerical columns to `object` type or skewing statistics.

#### Solution

Specify which values should be recognized as `NaN` (Not a Number) or `NaT` (Not a Time).

**Parameter to use**: `na_values`

```python
# Recognize a single custom string as missing
df = pd.read_csv("survey.csv", na_values='N/A')

# Recognize multiple custom strings and a number
df = pd.read_csv("measurements.csv", na_values=['?', 'Missing', -999])

# Column-specific missing values
df = pd.read_csv("experiment.csv", na_values={'ColA': ['---'], 'ColB': ['Invalid']})
```

### Problem 6: File Too Large to Fit in Memory

#### Problem

When trying to load a very large CSV (gigabytes), your script crashes with a `MemoryError`.

#### Solution

Process the file in smaller, manageable chunks rather than loading it all at once.

**Parameter to use**: `chunksize`

```python
chunk_size = 100000 # Read 100,000 rows at a time
csv_reader = pd.read_csv("huge_data.csv", chunksize=chunk_size)

# Example: Aggregate data without loading everything into RAM
total_sum = 0
for chunk in csv_reader:
    total_sum += chunk['Value_Column'].sum()
print(f"Total sum: {total_sum}")

# Example: Filter and save a subset of data
filtered_rows = []
for chunk in pd.read_csv("huge_data.csv", chunksize=chunk_size):
    filtered_chunk = chunk[chunk['Category'] == 'Important']
    filtered_rows.append(filtered_chunk)

final_filtered_df = pd.concat(filtered_rows)
final_filtered_df.to_csv("important_data.csv", index=False)
```

### Problem 7: Only Need Specific Columns

#### Problem

You have a CSV with many columns, but your analysis only requires a small subset. Loading all columns wastes memory and time.

#### Solution

Specify only the columns you need to load.

**Parameter to use**: `usecols`

```python
# Load specific columns by name
df = pd.read_csv("wide_dataset.csv", usecols=['ID', 'Feature_A', 'Target_Value'])

# Load specific columns by 0-based integer position
df = pd.read_csv("another_dataset.csv", usecols=[0, 5, 8]) # Load 1st, 6th, and 9th columns
```

### Problem 8: Incorrect Data Types After Loading

#### Problem

A column that should be numerical is `object` (string), or an integer column is `float` (e.g., `1.0` instead of `1`), or you want to optimize memory by using smaller integer/float types.

#### Solution

Explicitly define the data type for each column or use `converters` for complex transformations.

**Parameters to use**: `dtype`, `converters`

```python
# Specify dtypes for several columns for memory optimization and correctness
df = pd.read_csv(
    "large_numbers.csv",
    dtype={
        'UserID': 'int32',        # Smaller integer type
        'Rating': 'float16',      # Smaller float type
        'Category': 'category'    # Memory efficient for categorical data
    }
)

# Use a converter for complex cleaning (e.g., removing '$' and converting to float)
def clean_currency(value_str):
    return float(value_str.replace('$', ''))

df = pd.read_csv("financial_data.csv", converters={'Price': clean_currency})
```

### Problem 9: Malformed Rows (`ParserError` or Data Loss)

#### Problem

Some rows in your CSV have an inconsistent number of fields (too many or too few delimiters), causing `ParserError` or silently skipping data if not handled.

#### Solution

Control how pandas reacts to such malformed lines.

**Parameter to use**: `on_bad_lines`

```python
# Stop and raise an error on the first bad line (default, good for strict validation)
df = pd.read_csv("dirty_data.csv", on_bad_lines='error')

# Skip bad lines and print warnings (good for identifying issues while continuing)
df = pd.read_csv("dirty_data.csv", on_bad_lines='warn')

# Silently skip bad lines (use with caution, can hide problems)
df = pd.read_csv("dirty_data.csv", on_bad_lines='skip')
```

*   **Problem Solved**: Weird character errors, `UnicodeDecodeError`.
*   **Common Mistakes**: Not specifying `encoding` when dealing with files created on different operating systems or in different locales (e.g., Windows typically uses `cp1252` or `latin-1` for non-ASCII characters, while Linux/macOS often default to `UTF-8`).

### Parameter: `parse_dates`

#### Theory

*   **What it does**: The `parse_dates` parameter tells pandas to interpret specified columns as datetime objects. By default, pandas might load date columns as string (`object`) types.
*   **Why it is used**: Properly parsing dates into `datetime` objects is fundamental for any time-series analysis. It enables powerful time-based indexing, filtering (e.g., by month, year), and calculations (e.g., duration between events).
*   **Problem Solved**: Date column handling, enabling time-series analysis.
*   **Common Mistakes**: Not parsing dates, which leads to treating dates as strings, making time-based operations impossible or cumbersome. Pandas can also struggle with ambiguous date formats; specifying the format or providing a custom parser might be necessary for complex cases.

#### Syntax

```python
pd.read_csv("file.csv", parse_dates=['column_name'])
# or for multiple columns
pd.read_csv("file.csv", parse_dates=['col_A', 'col_B'])
# or for combined date/time columns
pd.read_csv("file.csv", parse_dates=[['Year', 'Month', 'Day']])
# or for boolean (tries to parse anything that looks like a date)
pd.read_csv("file.csv", parse_dates=True)
```

`parse_dates` can accept:
*   `bool`: If `True`, attempts to parse the index. If `False` (default), no parsing is done.
*   `list` of `int` or `str`: List of column numbers or names to parse.
*   `list` of `list`: For combining multiple columns into a single `datetime` column.

#### Theory

*   **What it does**: The `na_values` parameter allows you to specify a string or a list of strings that should be interpreted as `NaN` (Not a Number) or `NaT` (Not a Time) when reading the CSV file. By default, pandas recognizes common missing value representations like `''` (empty string), `'#N/A'`, `'#N/A N/A'`, `'#NA'`, `'-1.#IND'`, `'-1.#QNAN'`, `'-NaN'`, `'-nan'`, `'1.#IND'`, `'1.#QNAN'`, `'N/A'`, `'NA'`, `'NULL'`, `'NaN'`, `'n/a'`, `'nan'`, `'null'`.
*   **Why it is used**: Datasets often use various custom strings to denote missing data (e.g., `'?'`, `'-'`, `'None'`, `'N/A'`, a specific numerical value like `-999`). If these custom representations are not explicitly mapped to `NaN`, pandas will read them as regular strings or numbers, leading to incorrect analysis.
*   **Problem Solved**: Handling missing data, ensuring correct numerical/temporal column types, preventing errors in calculations.
*   **Common Mistakes**: Not accounting for custom missing value strings, which can lead to columns that should be numeric being read as `object` (string) type because of a non-numeric missing value indicator.

#### Syntax

```python
pd.read_csv("file.csv", na_values='string_to_replace')
# or for multiple strings
pd.read_csv("file.csv", na_values=['str1', 'str2'])
# or for column-specific missing values
pd.read_csv("file.csv", na_values={'col_A': ['str1'], 'col_B': ['str2', 'str3']})
```

`na_values` can accept:
*   `str` or `list-like`: A string or list of strings to recognize as `NaN`.
*   `dict`: A dictionary specifying per-column `NaN` values.

#### Example: Handling Custom Missing Values

````python
import pandas as pd
import numpy as np

# Create a dummy CSV with custom missing value indicators
data_na_custom_content = (
    'ID,Value1,Value2,Comments\n'
    '1,10,NA,Good\n'
    '2,?,20,Bad\n'
    '3,30,,Excellent\n'
    '4,-99,40,Fair'
)
with open('data_na_custom.csv', 'w') as f:
    f.write(data_na_custom_content)

# Read without specifying na_values
print("DataFrame without custom na_values:")
df_na_default = pd.read_csv('data_na_custom.csv')
print(df_na_default.dtypes)
display(df_na_default)

# Read with custom na_values for 'NA', '?', and -99
print("\nDataFrame with custom na_values (NA, ?, -99) specified:")
df_na_custom = pd.read_csv('data_na_custom.csv', na_values=['NA', '?', -99])
print(df_na_custom.dtypes)
display(df_na_custom)

# Clean up dummy file
os.remove('data_na_custom.csv')
````

#### Output Understanding

*   In `df_na_default`, `'NA'` and `'?'` in `Value1` and `Value2` columns are treated as strings. This forces `Value1` and `Value2` columns to be `object` (string) type, preventing numerical operations.
*   In `df_na_custom`, by providing `na_values=['NA', '?', -99]`, these specific strings/numbers are correctly converted to `NaN`. As a result, `Value1` and `Value2` are now `float64` (as `NaN` is a float representation of missing), allowing numerical operations. The empty string in `Value2` was also automatically recognized as `NaN` by pandas' default `na_values`.
