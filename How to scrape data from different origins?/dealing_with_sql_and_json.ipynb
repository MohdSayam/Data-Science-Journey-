# Working with SQL

### Working with SQLite (No Server Required)

Since you don't have an external MySQL server, we can use SQLite, which is a lightweight, file-based database. It's perfect for learning and development, as it doesn't require a separate server process and you can create an in-memory database or a database file directly in your Colab environment.

```python
import sqlite3

# Connect to an in-memory SQLite database
# This database will only exist for the duration of the current session
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

print("Connected to in-memory SQLite database.")
```

Now, let's create a table and insert some data using SQL commands.

```python
# Create a table
create_table_sql = """
CREATE TABLE employees (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    position TEXT NOT NULL,
    salary REAL
);
"""
cursor.execute(create_table_sql)

# Insert some data
insert_data_sql = """
INSERT INTO employees (name, position, salary) VALUES
('Alice Smith', 'Software Engineer', 75000.00),
('Bob Johnson', 'Project Manager', 90000.00),
('Charlie Brown', 'Data Analyst', 60000.00);
"""
cursor.execute(insert_data_sql)

# Commit the changes
conn.commit()

print("Table 'employees' created and data inserted.")
```

Finally, let's query the data we just inserted.

```python
# Query the data
select_data_sql = "SELECT * FROM employees;"
cursor.execute(select_data_sql)

rows = cursor.fetchall()

# Print the results
for row in rows:
    print(row)

# Close the connection (important for file-based databases, less critical for in-memory ones but good practice)
conn.close()

print("Connection closed.")
```

This demonstrates how to perform basic SQL operations within your notebook using SQLite, without needing an external database server. If you do eventually need to connect to an external MySQL server, you would replace `'localhost'` with the actual IP address or hostname of your MySQL server, and ensure it's configured to accept remote connections.
