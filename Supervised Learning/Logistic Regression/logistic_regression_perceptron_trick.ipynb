## Unveiling Logistic Regression: More Than Just a Line

Welcome to an exploration of Logistic Regression, a powerful classification algorithm often misunderstood due to its name. While 'regression' typically implies predicting continuous numbers (like a house's price), Logistic Regression is all about *classification* – predicting categories. Think of it as deciding between 'Spam' or 'Not Spam', or 'Class 0' and 'Class 1'. Our goal is to train a model that intelligently draws a clear boundary between these distinct groups.

### Why Linear Regression Falls Short for Classification

One might wonder, why not simply use a straight line from Linear Regression to separate our classes? Imagine trying to classify tumors as 'malignant' (1) or 'benign' (0) based on their size. Initially, a linear fit with a 0.5 threshold might seem perfect.

However, Linear Regression's sensitivity to outliers is its Achilles' heel here. Introduce a single, unusually large malignant tumor, and Linear Regression's underlying loss function will drastically adjust the entire line to accommodate this extreme point. This shift can lead to misclassifying perfectly benign cases, highlighting a fundamental flaw: we don't need a model that's influenced by *how far* a correct point is; we simply need a distinct dividing line.

### The Perceptron Trick: A Geometric Intuition

Before Logistic Regression, the **Perceptron** offered a brilliant geometric solution for drawing this boundary through trial and error. Visualize a 2D plot with two classes of data points (e.g., red for Class 0, blue for Class 1):

1.  **The Random Start:** Begin by drawing an arbitrary straight line across your data.
2.  **The Test:** Select a random data point and check which side of the line it falls on.
3.  **The Trick (Correction):**
    *   If the line correctly classifies the point, do nothing.
    *   If the line makes a mistake, metaphorically *grab* the line and *pull* it towards the misclassified point, ushering it into its correct territory.

### The Math Behind the 'Pull': The Update Function

Translating this 'physical pull' into mathematics involves a remarkably simple update rule for our weights ($W$), using the coordinates of the misclassified point ($X$). This is the **Perceptron Update Rule**:

$$W_{\text{new}} = W_{\text{old}} + \text{learning\_rate} \cdot (y - \hat{y}) \cdot X$$

Let's dissect the core of this engine: $(y - \hat{y})$. Here, $y$ is the true label and $\hat{y}$ is the line's prediction.

*   **Scenario A: Correct Classification**
    *   If $y=1, \hat{y}=1$, then $(1-1)=0$.
    *   If $y=0, \hat{y}=0$, then $(0-0)=0$.
    *   **Result:** The update is multiplied by zero, leaving the weights (and thus the line) unchanged. No correction needed.

*   **Scenario B: False Negative (True label is 1, predicted 0)**
    *   If $y=1, \hat{y}=0$, then $(1-0)=+1$.
    *   **Result:** We *add* the point's coordinates ($X$) to our weights. Geometrically, adding this vector pulls the line *towards* the misclassified point.

*   **Scenario C: False Positive (True label is 0, predicted 1)**
    *   If $y=0, \hat{y}=1$, then $(0-1)=-1$.
    *   **Result:** We *subtract* the point's coordinates ($X$) from our weights. Geometrically, subtracting this vector pushes the line *away* from the misclassified point.

Repeated thousands of times in a loop, this process causes the line to 'bounce' and adjust, eventually settling into the optimal gap that separates the two classes.

### The Perceptron's Limitation: Why We Need Transformations

The Perceptron's ingenuity has one critical drawback: it uses a **Step Function**. It delivers a rigid, binary output (a hard 0 or a hard 1), lacking any sense of probability or nuance. In real-world applications, knowing "I am 80% sure this is a tumor" or "I am 51% sure" is vital for safe and informed decision-making. The Perceptron cannot provide these probabilities.

To overcome this, we need a mathematical transformation. Instead of an abrupt step, we require a function that can *squish* any raw input into a smooth probability curve. This transition sets the stage for our next lesson: replacing the Step Function with the **Sigmoid Function** to evolve this bouncing Perceptron line into true Logistic Regression.

```python
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt

X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
def perceptron(X, y):
  X = np.insert(X, 0, 1, axis=1)
  weights = np.ones(X.shape[1])
  lr = 0.1

  for i in range(1000):
    j = np.random.randint(0, 100)
    y_hat = step(np.dot(X[j], weights))
    weights = weights + lr * (y[j] - y_hat) * X[j]

  return weights[0], weights[1:]

```

```python
def step(z):
  return 1 if z>0 else 0
```

```python
intercept_, coef_ = perceptron(X, y)
print(intercept_)
print(coef_)
```

```python
m = -(coef_[0] / coef_[1])
b = -(intercept_ / coef_[1])
```

```python
def perceptron(X,y):

    m = []
    b = []

    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(200):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]

        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))

    return m,b
```

```python
m, b = perceptron(X, y)
```
