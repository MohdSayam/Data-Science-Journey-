### 2. Line Plot

MOST BASIC PLOT.

Used for:

- trends
- time series
- growth analysis

#### Your First Plot

```python
# Define data for the x and y axes
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

# Create a line plot using plt.plot()
plt.plot(x, y)

# Display the plot using plt.show()
plt.show()
```

IMPORTANT:

`plt.show()` displays the graph.

Understanding:

- `x` → horizontal axis
- `y` → vertical axis

#### Add Title and Labels

VERY IMPORTANT. ALWAYS label graphs properly.

```python
# Using the same data x and y from above

# Create a line plot
plt.plot(x, y)

# Add a title to the plot
plt.title("Sales Growth")

# Add a label for the x-axis
plt.xlabel("Months")

# Add a label for the y-axis
plt.ylabel("Revenue")

# Display the plot
plt.show()
```

#### Add Markers

```python
# Using the same data x and y from above

# Create a line plot with circular markers ('o') at each data point
plt.plot(x, y, marker="o")

# Add title and labels for clarity
plt.title("Sales Growth with Markers")
plt.xlabel("Months")
plt.ylabel("Revenue")

# Display the plot
plt.show()
```

#### Line Styles

```python
# Using the same data x and y from above

# Create a line plot with a dashed line style ('--')
plt.plot(x, y, linestyle="--")

# Add title and labels
plt.title("Sales Growth with Dashed Line")
plt.xlabel("Months")
plt.ylabel("Revenue")

# Display the plot
plt.show()
```

### 3. Scatter Plot

SUPER IMPORTANT in Data Science.

Used for:

- relationship analysis
- correlation
- distributions
- clustering intuition

```python
# Define data for x and y axes
x = [1, 2, 3, 4, 5]
y = [10, 15, 14, 20, 25]

# Create a scatter plot using plt.scatter()
plt.scatter(x, y, marker='^')
plt.grid(True)

# Display the plot
plt.show()
```

### 4. Bar Chart

VERY common in analytics.

Used for:

- category comparison
- counts
- frequency

```python
# Define data for student names and their marks
students = ["Aman", "Riya", "Karan"]
marks = [85, 92, 78]

# Create a vertical bar chart using plt.bar()
plt.bar(students, marks)

# Add title and labels
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

# Display the plot
plt.show()
```

#### Real Usage

Used for:

### 5. Histogram

EXTREMELY IMPORTANT.

Used for:

VERY important in ML.

```python
# Define a list of marks
marks_data = [55, 60, 65, 70, 75, 80, 85, 90, 95, 62, 78, 83, 71, 88, 92]

# Create a histogram using plt.hist()
plt.hist(marks_data)

# Add title and labels
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# Display the plot
plt.show()
```

### 6. Pie Chart

Less common in ML.
Still useful sometimes.

```python
# Define labels for each slice
labels = ["IT", "HR", "Finance"]
# Define the size (proportion) of each slice
sizes = [40, 35, 25]

# Create a pie chart using plt.pie()
plt.pie(sizes, labels=labels)

# Add a title
plt.title("Department Staff Distribution")

# Ensure the circle is drawn proportionally
plt.axis('equal')

# Display the plot
plt.show()
```

#### Add Percentages

```python
# Using the same labels and sizes data

# Create a pie chart with percentages displayed on each slice
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%" # Format for displaying percentage values
)

# Add a title
plt.title("Department Staff Distribution with Percentages")

# Ensure the circle is drawn proportionally
plt.axis('equal')

# Display the plot
plt.show()
```

### 7. Figure Size

VERY IMPORTANT.

```python
# Using the same x and y data from the line plot example
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

# Create a figure with a specific size (width=8 inches, height=5 inches)
plt.figure(figsize=(8, 5))

# Create a line plot on this figure
plt.plot(x, y)

# Add title and labels
plt.title("Sales Growth (Custom Figure Size)")
plt.xlabel("Months")
plt.ylabel("Revenue")

# Display the plot
plt.show()
```

Format:

`(width, height)`

#### Why Important?

Without proper figure size:

- plots become unreadable
- labels overlap

### 8. Grid

Makes graphs easier to read.

```python
# Using the same x and y data

# Create a line plot
plt.plot(x, y)

# Add a title and labels
plt.title("Sales Growth with Grid")
plt.xlabel("Months")
plt.ylabel("Revenue")

# Add a grid to the plot
plt.grid(True)

# Display the plot
plt.show()
```

### 9. Legends

Used when multiple plots exist.

```python
# Define data for x and two y-series
x_series = [1, 2, 3, 4]

y1 = [10, 20, 30, 40] # Data for Product A
y2 = [15, 25, 35, 45] # Data for Product B

# Plot the first series with a label
plt.plot(x_series, y1, label="Product A", marker='o')

# Plot the second series with a label
plt.plot(x_series, y2, label="Product B", marker='x')

# Add title and labels
plt.title("Product Sales Comparison")
plt.xlabel("Quarter")
plt.ylabel("Sales ($)")

# Display the legend to identify each line
plt.legend()

# Add a grid for readability
plt.grid(True, linestyle='--')

# Display the plot
plt.show()
```

#### Real Usage

Compare:

- multiple products
- models
- trends
- companies

### 10. Multiple Plots Together (Subplots)

```python
# Using the x_series, y1, and y2 data from the legends example

# Create a figure with 1 row, 2 columns, and select the first subplot
plt.subplot(1, 2, 1)
plt.plot(x_series, y1, color='blue')
plt.title("Product A Sales")
plt.xlabel("Quarter")
plt.ylabel("Sales ($)")
plt.grid(True)

# Select the second subplot
plt.subplot(1, 2, 2)
plt.plot(x_series, y2, color='green')
plt.title("Product B Sales")
plt.xlabel("Quarter")
plt.ylabel("Sales ($)")
plt.grid(True)

# Adjust layout to prevent overlapping titles/labels
plt.tight_layout()

# Display the plot
plt.show()
```

Format:

`(rows, columns, position)`

#### Better Modern Method (Object-Oriented)

VERY IMPORTANT.

```python
# Using the x_series and y1 data

# Create a figure and a set of subplots (axes)
# figsize sets the overall size of the figure
fig, ax = plt.subplots(figsize=(6, 4))

# Plot data on the created axes object (ax)
ax.plot(x_series, y1, color='purple', marker='s')

# Set the title for the axes
ax.set_title("Sales (Object-Oriented Method)")

# Set labels for the x and y axes
ax.set_xlabel("Quarter")
ax.set_ylabel("Sales ($)")

# Add a grid to the axes
ax.grid(True)

# Display the plot
plt.show()
```

#### Why This Method Matters

Professional plotting uses:

- object-oriented approach

More scalable and cleaner.

### Common Beginner Mistakes

- **Forgetting `plt.show()`**
  - Graph won’t display sometimes.

- **Wrong x/y lengths**
  - Example: `x = [1,2,3]`, `y = [1,2]` leads to an Error.
  - Lengths must match.

- **Too Many Labels**
  - Graph becomes unreadable.

- **Not Using Titles**
  - Always label plots properly.

Create subplots:

```python
# Task 6: Subplots - one histogram, one bar chart

# Data for histogram (e.g., student scores)
student_scores = [65, 70, 72, 75, 80, 81, 85, 88, 90, 92, 95, 98, 60, 78, 83, 86, 91]

# Data for bar chart (e.g., departmental budget)
departments = ['Sales', 'Marketing', 'R&D', 'HR']
budgets = [50, 45, 70, 30] # in thousands

# Create a figure and a set of subplots (1 row, 2 columns)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Subplot 1: Histogram
axes[0].hist(student_scores, bins=5, edgecolor='black', color='teal', alpha=0.7)
axes[0].set_title('Distribution of Student Scores')
axes[0].set_xlabel('Score')
axes[0].set_ylabel('Number of Students')
axes[0].grid(axis='y', linestyle='--')

# Subplot 2: Bar Chart
axes[1].bar(departments, budgets, color=['orange', 'green', 'blue', 'red'])
axes[1].set_title('Departmental Budgets')
axes[1].set_xlabel('Department')
axes[1].set_ylabel('Budget (Thousands $)')
axes[1].grid(axis='y', linestyle='--')

# Adjust layout to prevent overlapping elements
plt.tight_layout()

# Display the plots
plt.show()
```
