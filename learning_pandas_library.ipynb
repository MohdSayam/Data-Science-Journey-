## 1. Import Pandas

## 2. What is a Series?

A Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floats, Python objects, etc.). You can think of it as a single column of data with an associated index.

### Creating a Series

````python
# Create a simple Series with default integer index
s = pd.Series([10, 20, 30, 40])
print(s)

print(pd.Series([1,2,9,0]))
````

### Custom Index

You can also define a custom index for your Series, which can be useful for labeling data points.

````python
# Create a Series with a custom string index
s_custom_index = pd.Series([10,9,7.5], index=['Maths', 'English', 'MLF'])
print(s_custom_index)
````

### Accessing Values

You can access values in a Series using its index.

````python
# Access a value using its label
print(f"Math CGPA: {s_custom_index['Maths']}")

# Access a value using its integer position (like a list)
# nowadays pandas prefer .loc when we want by label/name and .iloc when we want by integer so it is not good to use without them
print(f"First score: {s_custom_index.iloc[0]}")
````

## 3. What is a DataFrame?

The DataFrame is the most important Pandas object. It's a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). Think of it like an Excel sheet, a SQL table, or a dataset.

### Creating Your First DataFrame

DataFrames are commonly created from dictionaries, where keys become column names and values are lists of data for each column.

````python
# Create a DataFrame from a dictionary
data = {
    "Name": ["Aman", "Riya", "Karan"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)
print(df)
````

### Important Terminology

*   **Rows:** Horizontal entries, identified by an index (e.g., 0, 1, 2 in the example above).
*   **Columns:** Vertical entries, identified by column names (e.g., 'Name', 'Age', 'Marks').

## 4. Basic Dataset Inspection

Inspecting your data is often the very first step in any data analysis workflow. These methods help you get a quick overview of your DataFrame's structure and content.

### `head()`

Shows the first `n` rows (default is 5). Useful for a quick peek at the data.

````python
# Display the first 5 rows
print("df.head():\n", df.head())

# Display the first 2 rows
print("\ndf.head(2):\n", df.head(2))
````

### `tail()`

Shows the last `n` rows (default is 5). Useful for checking the end of the data, especially after sorting or appending.

````python
# Display the last 5 rows
print("df.tail():\n", df.tail())

# Display the last 1 row
print("\ndf.tail(1):\n", df.tail(1))
````

### `shape`

Returns a tuple representing the dimensionality of the DataFrame (rows, columns). This is very important for understanding the size of your dataset.

````python
# Get the number of rows and columns
print("df.shape:", df.shape)
print(f"Meaning: {df.shape[0]} rows and {df.shape[1]} columns")
````

### `columns`

Returns a list of all column names in the DataFrame.

````python
# Get column names
print("df.columns:", df.columns)
````

### `info()`

This method prints a concise summary of a DataFrame, including the index dtype and column dtypes, non-null values, and memory usage. This is super important for quickly checking data types and identifying missing values.

````python
# Get a summary of the DataFrame
df.info()
````

### `describe()`

Generates descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset's distribution. By default, this only includes numerical columns.

````python
# Get descriptive statistics for numerical columns
print("df.describe():\n", df.describe())
````

## 5. Selecting Columns

DataFrames allow flexible ways to select one or more columns.

### Single Column

Selecting a single column returns a Pandas Series.

````python
# Select the 'Name' column
name_series = df["Name"]
print("Type of df['Name']:", type(name_series))
print(name_series)
````

### Multiple Columns

To select multiple columns, pass a list of column names. This will return a DataFrame.

````python
# Select 'Name' and 'Marks' columns
name_marks_df = df[["Name", "Marks"]]
print("Type of df[['Name', 'Marks']]:", type(name_marks_df))
print(name_marks_df)
````

### Important Difference

*   **Single brackets `df["Column"]`**: Returns a `Series`.
*   **Double brackets `df[["Column"]]`**: Returns a `DataFrame` (even if it's just one column). This distinction is very important for subsequent operations.

## 6. Selecting Rows

Pandas provides two primary methods for selecting rows:

*   **`.loc`**: Label-based selection. Used for selecting by row and column labels.
*   **`.iloc`**: Integer-location based selection. Used for selecting by integer positions.

### `loc` → Label-based

Use `loc` to select rows by their index labels (which are 0, 1, 2 by default in our DataFrame).

````python
# Select the row with index label 0
print("df.loc[0]:\n", df.loc[0])
````

### `iloc` → Index-position based

Use `iloc` to select rows by their integer positions (0-indexed).

````python
# Select the row at integer position 0
print("df.iloc[0]:\n", df.iloc[0])
````

### Selecting Specific Rows & Columns

You can combine `loc` or `iloc` to select specific rows and columns simultaneously.

````python
# Select 'Name' from the row with label 0
print("df.loc[0, 'Name']:", df.loc[0, "Name"])

# Select 'Age' from the row at integer position 2
print("df.iloc[2, 1]:", df.iloc[2, 1]) # 2nd row (Karan), 1st column (Age)
````

### Multiple Rows

Both `loc` and `iloc` support slicing for multiple rows.

````python
# Select rows with labels 0 to 1 (inclusive)
print("df.loc[0:1]:\n", df.loc[0:1])

# Select rows at integer positions 0 to 1 (exclusive of 2)
print("\ndf.iloc[0:2]:\n", df.iloc[0:2])
````

### Important Difference from Python Slicing

When slicing with `loc` using `df.loc[start:end]`, **both** the `start` and `end` labels are **inclusive**. This is different from standard Python list slicing, where the `end` is exclusive.

## 7. Filtering Data (Most Important Skill)

Filtering is a fundamental operation in data science. It allows you to select a subset of your data based on one or more conditions. This is often called boolean indexing.

### Example: Single Condition

Let's filter for students who scored more than 80 marks.

````python
# Filter for students with Marks > 80
students_high_marks = df[df["Marks"] > 80]         # df[] will give all then inside another df['Marks'] which will give marks and we are checking things there
print("Students with Marks > 80:\n", students_high_marks)
````

### Multiple Conditions

You can combine multiple conditions using logical operators (`&` for AND, `|` for OR, `~` for NOT). Remember to enclose each condition in parentheses.

````python
# Filter for students with Marks > 80 AND Age > 20
filtered_df = df[(df['Marks'] > 80) & (df['Age'] > 20)]
print("Students with Marks > 80 and Age > 20:\n", filtered_df)
````

### Real Data Science Usage

Filtering is used extensively for tasks like:

*   Identifying specific types of transactions (e.g., fraud).
*   Selecting customers based on demographics or behavior.
*   Analyzing data for specific time periods or categories.
*   Handling outliers or anomalies.

## 8. Adding New Columns

Adding new columns to a DataFrame is a very common operation, often used to derive new features or categories from existing data.

### Example: Boolean Column

Let's add a 'Passed' column based on whether 'Marks' are above 40.

````python
# Add a 'Passed' column
df["Passed"] = df["Marks"] > 40
df["Good_Student"] = df["Marks"] > 80
print("DataFrame with 'Passed' column:\n", df)
````

## 9. Updating Data

Modifying existing data in a DataFrame is crucial for data cleaning and transformation.

### Example: Update a Single Value

Use `.loc` to update a specific cell by its row label and column label.

````python
# Update Aman's marks to 95
df.loc[0, "Marks"] = 95
df.loc[2, "Age"] = 19
print("DataFrame after updating Aman's Marks:\n", df)
````

## 10. Sorting Data

Sorting allows you to reorder your DataFrame rows based on the values in one or more columns.

````python
# Sort by 'Marks' in ascending order
print("Sorted by Marks (ascending):\n", df.sort_values("Marks"))
````

Use the `ascending=False` parameter for descending order.

## 11. Missing Values (Critical)

Real-world datasets almost always contain missing values. Handling them properly is a critical step in data preprocessing.

### Example DataFrame with Missing Values

Missing values are often represented as `None` or `NaN` (Not a Number).

````python
import numpy as np

missing_data = {
    "Name": ["Aman", "Riya", None, "Deepak"],
    "Marks": [90, np.nan, 70, 88],
    "City": ["Delhi", "Mumbai", "Pune", None]
}

df_missing = pd.DataFrame(missing_data)
print("Original DataFrame with missing values:\n", df_missing)
````

### Detect Missing Values

The `isnull()` method returns a boolean DataFrame indicating where values are missing.

````python
# Detect missing values
print("\nMissing values (boolean DataFrame):\n", df_missing.isnull())
````

### Count Missing Values

Combining `isnull()` with `sum()` is a very common way to count missing values per column.

````python
# Count missing values per column
print("\nCount of missing values per column:\n", df_missing.isnull().sum())
````

### Drop Missing Values

The `dropna()` method removes rows or columns with missing values. By default, it drops rows that contain *any* missing values.

````python
# Drop rows with any missing values
df_dropped = df_missing.dropna()
print("\nDataFrame after dropping rows with missing values:\n", df_dropped)
````

### Fill Missing Values

The `fillna()` method allows you to replace missing values with a specified value (e.g., 0, mean, median, mode, or a constant string).

````python
# Fill numerical missing values with 0 and categorical missing values with 'Unknown'
df_filled = df_missing.copy()
df_filled['Marks'] = df_filled['Marks'].fillna(0) # Fill numerical column
df_filled['Name'] = df_filled['Name'].fillna('Unknown') # Fill object column
df_filled['City'] = df_filled['City'].fillna('Unknown') # Fill object column
print("\nDataFrame after filling missing values:\n", df_filled)
````

### Real ML Usage

In machine learning, you never ignore missing values. Common strategies include:

*   **Removal:** Dropping rows/columns if missing data is sparse or the column is not important.
*   **Imputation:** Filling missing values using statistical methods (mean, median, mode) or more advanced techniques (e.g., regression imputation, K-Nearest Neighbors imputation).
*   **Prediction:** Treating missing values as a target variable in a separate model.

`df.groupby(["Department", "City"])["Salary"].mean()`

IMPORTANT.

Groupby often creates hierarchical index.

`result = df.groupby("Department")["Salary"].mean()`

`print(result.reset_index())`

## 8. `merge()` — VERY IMPORTANT

Equivalent to SQL JOIN.

Used constantly.

```python
employees = pd.DataFrame({
```

```python
salaries = pd.DataFrame({

```
### Merge
`pd.merge(employees, salaries, on="ID")`

**Output:**

```
   ID   Name  Salary
0   1   Aman   60000
1   2   Riya   50000
2   3  Karan   70000
```

### Why Important?

Real data comes from:

*   multiple CSV files
*   multiple databases
*   APIs
*   tables

You combine datasets using merges.

### Types of Joins

*   **Inner Join** (default)
    `pd.merge(a, b, on="ID", how="inner")`
*   **Left Join**
    `pd.merge(a, b, on="ID", how="left")`
*   **Right Join**
    `pd.merge(a, b, on="ID", how="right")`
*   **Outer Join**
    `pd.merge(a, b, on="ID", how="outer")`

VERY IMPORTANT for interviews too.

````python
employees = pd.DataFrame({
    "ID": [1,2,3],
    "Name": ["Aman", "Riya", "Karan"]
})
salaries = pd.DataFrame({
    "ID": [1,2,3],
    "Salary": [60000, 50000, 70000]
})
pd.merge(employees, salaries, on="ID")

# Example of a left join with a missing ID in 'salaries'
salaries_missing = pd.DataFrame({
    "ID": [1, 3],
    "Salary": [60000, 70000]
})
pd.merge(employees, salaries_missing, on="ID", how="outer")
````
